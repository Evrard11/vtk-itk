# Visualisation des masques de segmentation tumorale

```python
import itk
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

gre1  = itk.imread('data/case6_gre1.nrrd', itk.F)
gre2  = itk.imread('data/case6_gre2_registered.nrrd', itk.F)
mask1 = itk.imread('data/mask_gre1.nrrd')
mask2 = itk.imread('data/mask_gre2.nrrd')

arr1 = itk.array_from_image(gre1)
arr2 = itk.array_from_image(gre2)
m1   = itk.array_from_image(mask1)
m2   = itk.array_from_image(mask2)

vol1 = (m1 > 0).sum()
vol2 = (m2 > 0).sum()
print(f'Volume GRE1 : {vol1} mm3')
print(f'Volume GRE2 : {vol2} mm3')
print(f'Delta        : {vol2 - vol1:+d} mm3')

z_tumor = [z for z in range(m1.shape[0]) if m1[z].sum() > 0]
print(f'Coupes avec tumeur : z={z_tumor[0]} a z={z_tumor[-1]} ({len(z_tumor)} coupes)')
```

```python
# Masque GRE1 superpose sur l'image originale (8 coupes regulierement espacees)
vmax = np.percentile(arr1, 99)
sample = z_tumor[::max(1, len(z_tumor) // 8)][:8]

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, z in zip(axes.flat, sample):
    ax.imshow(arr1[z], cmap='gray', vmin=0, vmax=vmax)
    ax.imshow(m1[z], cmap='Reds', alpha=0.5)
    ax.set_title(f'z={z}')
    ax.axis('off')
plt.suptitle('Masque GRE1 (rouge) sur GRE1')
plt.tight_layout()
plt.show()
```

```python
# Comparaison GRE1 vs GRE2 sur la coupe centrale : evolution de la tumeur entre les deux temps
z_center = z_tumor[len(z_tumor) // 2]
vmax2 = np.percentile(arr2, 99)

fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(arr1[z_center], cmap='gray', vmin=0, vmax=vmax)
axes[0].imshow(m1[z_center], cmap='Reds', alpha=0.4)
axes[0].set_title(f'GRE1  z={z_center}')

axes[1].imshow(arr2[z_center], cmap='gray', vmin=0, vmax=vmax2)
axes[1].imshow(m2[z_center], cmap='Blues', alpha=0.4)
axes[1].set_title(f'GRE2 (recale)  z={z_center}')

axes[2].imshow(arr1[z_center], cmap='gray', vmin=0, vmax=vmax)
axes[2].imshow(m1[z_center], cmap='Reds', alpha=0.4)
axes[2].imshow(m2[z_center], cmap='Blues', alpha=0.3)
axes[2].set_title('Superposition (rouge=GRE1, bleu=GRE2)')

for ax in axes:
    ax.axis('off')
plt.tight_layout()
plt.show()
```
